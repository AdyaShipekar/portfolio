**Strengths and Weaknesses:**

Strengths
1. I can trace code step by step to predict its output, including swapping two variables with a temporary variable, following assignment statements in order, and tracking values through loops and list searches.
2. I can use and combine procedures as abstractions, such as calling built-in procedures like MAX or writing code around a given procedure, which lets me solve problems without needing to know how each procedure works internally.
3. I can interpret data and metadata to draw conclusions, including reading trends from graphs, forming hypotheses from data sets, and understanding how metadata like dates and locations helps organize photos and files.

Weaknesses
1. I struggle to tell whether two algorithms produce the same result, especially when small details differ, like whether a sum is reset inside or outside a loop or when the division to find an average happens.
2. I have trouble judging how an algorithm's runtime grows as the input gets larger, such as recognizing when a process that checks every combination of items becomes too slow to be practical.
3. I don't fully understand undecidable problems, particularly that no algorithm can always determine whether any given program will stop or run forever in an infinite loop.




### **Class Retrospective**

I currently have an A in this class, but my grade doesn't fully reflect how much I still have to learn. This first stretch has shown me what I'm good at and where I need to grow, both technically and as a teammate.

**What's Going Well**
- **Owning my tools.** My development environment has gone from something I used by following instructions to something I can troubleshoot on my own. Fixing my broken `make` setup meant I could test changes locally in seconds instead of waiting for a GitHub Pages deploy. Organizing my repos into separate `opencs` and personal directories also ended the "Git isn't working" moments that came from running commands in the wrong folder.
- **Persistence when things break.** When the Python coderunner wouldn't execute, I opened the code, found that it was missing a Python workspace, and kept debugging until the prompts ran. It was the first time I edited tooling instead of just using it. When pages failed to render, I isolated changes one at a time instead of reverting everything.
- **Helping others.** Setting up tools for Joan, Ainsley, and Jailene on both Windows and Mac pushed me to understand how install paths and shell commands differ between operating systems. Explaining the full setup process, from installing WSL to forking, cloning, and opening VS Code, is what proved to me that I actually understood it.
- **Leading as Scrum Master.** I broke our CPT concept into tasks, confirmed each person understood their piece, and checked in mid-sprint so problems came up early instead of the night before a checkpoint.

**What I Need to Improve**
- **Coding fundamentals.** My JavaScript and coding challenges are at a 3, and my AP practice exam showed specific gaps: comparing whether two algorithms produce the same result, judging how runtime grows with input size, and understanding undecidable problems. I'm strong at setting up tools, but I need to put the same effort into writing and reasoning about code.
- **Using AI with intention.** I mostly used AI to explain and debug errors, and I learned that vague prompts gave me code I couldn't understand or fix. Going forward, I want to give AI full context and make sure I can explain anything it produces before I use it.
- **Design and layout.** My portfolio works, but its theme and layout are at a 3. I want my page to look intentional, not just functional.
- **Seeing myself accurately.** In some areas, like attendance, communication, mentoring, and organization, my peers didn't rate me the same as I rated myself. I want to find out what they saw so I'm judging myself by how my work comes across to others, not just by my own effort.
- **Engagement and asking for help.** I'm comfortable helping others, but I can be more active in asking questions and contributing in class discussions rather than only stepping in once something is broken.

**Goals for Next Sprint**
1. Practice algorithm comparison and efficiency problems until I can explain why two code segments do or don't produce the same output.
2. Complete more JavaScript challenges on my own before turning to AI, and only use AI-generated code I can fully explain.
