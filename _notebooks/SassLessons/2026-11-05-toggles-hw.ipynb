## Toggles

A toggle is an on/off switch. The checkbox inside it stores the state, `:checked` reads that state in SCSS, and the track changes color to show it. JavaScript reads the same state with `.checked`.

### Popcorn Hack: Fill in the Blanks

1. The **checkbox** on a toggle keeps track of whether it is on or off.
2. The **slider** is the visible part of the toggle that changes color when it is turned on.
3. The `.switch` class controls the toggle's position, size, padding, and **alignment**.
4. The **border-radius** property makes the slider have rounded edges.
5. The `&:before` class dictates the **white circle** of the toggle.

### Popcorn Hack: Find the Bug

The toggle was missing its track span, so there was nothing visible to click. Adding `class="ocs__toggle-track"` creates the track, and `aria-hidden="true"` tells screen readers to skip it because it is only decoration.

```python
%%html
<!-- UI_RUNNER: Toggles Find the Bug | show_code: true -->
<label class="ocs__toggle ocs__toggle--green">
  <input id="bug-toggle" class="ocs__toggle-input" type="checkbox">
  <!-- This was the missing line: the track is the part you actually see -->
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Toggle works! 🎉</span>
</label>
```

### Popcorn Hack: Connect to Something Real

This toggle shows and hides an image on the page. The toggle and the image each get an ID, and JavaScript connects them.

```python
%%html
<!-- UI_RUNNER: Toggles Connect | show_code: true -->
<label class="ocs__toggle ocs__toggle--green">
  <input id="image-toggle" class="ocs__toggle-input" type="checkbox" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show Image</span>
</label>

<div>
  <img id="my-image" src="https://upload.wikimedia.org/wikipedia/commons/thumb/9/96/Sass_Logo_Color.svg/320px-Sass_Logo_Color.svg.png" width="200" alt="Sass logo">
</div>

<script>
  (() => {
    const toggle = document.getElementById("image-toggle");
    const image = document.getElementById("my-image");

    // hidden is the opposite of checked: toggle on means image shown
    toggle.addEventListener("change", () => {
      image.hidden = !toggle.checked;
    });
  })();
</script>
```

### Homework: Study Session Settings

A settings panel for a study page. There are four toggles, and each one controls a real feature on the card below it. A counter at the top shows how many toggles are on and updates every time one flips.

```python
%%html
<!-- UI_RUNNER: Toggles Homework | show_code: true -->
<div id="study-settings">
  <!-- aria-live="polite" makes screen readers announce the new count when it changes -->
  <p><strong>Toggles on:</strong> <output id="study-count" aria-live="polite">0</output> / 4</p>

  <!-- Each toggle uses the same four-part ocs__toggle structure: label, input, track, text -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="study-notes" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show Notes</span>
  </label>
  <br>
  <label class="ocs__toggle" style="--ocs-toggle-on: #FACC15">
    <input id="study-highlight" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Highlight Key Terms</span>
  </label>
  <br>
  <label class="ocs__toggle" style="--ocs-toggle-on: #3B82F6">
    <input id="study-large" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Large Text</span>
  </label>
  <br>
  <label class="ocs__toggle" style="--ocs-toggle-on: #8B5CF6">
    <input id="study-focus" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Focus Mode</span>
  </label>

  <div id="study-card" style="margin-top: 1rem; padding: 1rem; border: 1px solid #6B7280; border-radius: 0.5rem; transition: all 0.3s;">
    <h4 style="margin-top: 0;">SASS Toggles</h4>
    <p id="study-body">
      A <span class="study-term">toggle</span> wraps a hidden <span class="study-term">checkbox</span>.
      The <span class="study-term">:checked</span> selector changes the look of the
      <span class="study-term">track</span> when it is on.
    </p>
    <ul id="study-notes-list" hidden>
      <li>The input is invisible (opacity 0) but still holds the state.</li>
      <li><code>input:checked + .ocs__toggle-track</code> works because the track comes right after the input.</li>
      <li>JavaScript reads the same state with <code>.checked</code>.</li>
    </ul>
    <p id="study-status" style="font-size: 14px; opacity: 0.8;"></p>
  </div>
</div>

<script>
  // Wrapped in an arrow function so these names don't clash with the other cells on the page
  (() => {
    const notes = document.getElementById("study-notes");
    const highlight = document.getElementById("study-highlight");
    const large = document.getElementById("study-large");
    const focus = document.getElementById("study-focus");

    const count = document.getElementById("study-count");
    const card = document.getElementById("study-card");
    const body = document.getElementById("study-body");
    const notesList = document.getElementById("study-notes-list");
    const status = document.getElementById("study-status");
    const terms = card.querySelectorAll(".study-term");

    // Scoped to #study-settings so toggles in the other cells don't get counted
    const allToggles = document.querySelectorAll("#study-settings .ocs__toggle-input");

    function update() {
      // I thought .checked was interesting because it lets JavaScript tell
      // whether the toggle is currently turned on or off, the same state :checked reads in SCSS
      notesList.hidden = !notes.checked;

      // Selection: IF the toggle is on, apply the active style, ELSE remove it
      terms.forEach((term) => {
        term.style.background = highlight.checked ? "#FACC15" : "transparent";
        term.style.color = highlight.checked ? "#000000" : "inherit";
      });

      card.style.fontSize = large.checked ? "1.25rem" : "1rem";

      if (focus.checked) {
        card.style.borderColor = "#8B5CF6";
        card.style.boxShadow = "0 0 0 3px rgba(139, 92, 246, 0.35)";
      } else {
        card.style.borderColor = "#6B7280";
        card.style.boxShadow = "none";
      }

      // Array.from turns the NodeList into an array so filter can count the checked ones
      const onCount = Array.from(allToggles).filter((t) => t.checked).length;
      count.textContent = onCount;

      const labels = Array.from(allToggles)
        .filter((t) => t.checked)
        .map((t) => t.parentElement.querySelector(".ocs__toggle-label").textContent);
      status.textContent = onCount === 0 ? "No settings on." : "Active: " + labels.join(", ");
    }

    // "change" fires every time a toggle is flipped, by mouse or keyboard
    allToggles.forEach((t) => t.addEventListener("change", update));
    update();
  })();
</script>
```

## Reflection

- The checkbox is still there, it is just invisible. Clicking the label flips it because the input is inside the label.
- `input:checked + .ocs__toggle-track` only works because the track comes directly after the input. If the order changed, the sibling selector would stop matching.
- I gave the on color a CSS variable (`--ocs-toggle-on`) so one class handles every color, and I override it inline instead of writing a new class per color.
- The counter uses the same `.checked` property as the features, so the page always agrees with what the toggles show.
