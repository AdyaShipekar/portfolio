## Design Thinking

**Empathize:** A veteran or volunteer may be filling this out on a phone, so the form should not make them guess what a field wants.

**Define:** Every field needs a clear purpose and a size that fits its answer, with no custom styling.

**Ideate:** I wrote the signup labels as questions, so the placeholder stays a hint instead of doing the job of a label.

**Prototype:** Size follows the expected answer: small for a name, medium for an email, large for a category. `gradient` appears once per form, on the field that matters most.

**Test:** Each input has a label tied to it with `for` and `id`, and the email field uses `type="email"` so phones show the right keyboard.
