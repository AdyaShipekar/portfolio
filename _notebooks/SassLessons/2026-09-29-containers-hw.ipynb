## Design Thinking

**Empathize:** Families check parade details on their phones, often while already on the way, so the key facts need to be easy to find.

**Define:** The hub needs one clear structure: a single container, a single card, and no layout code that breaks on small screens.

**Ideate:** I treated the page as a reading order: what the event is, when and where it is, who is in it, and one last tip.

**Prototype:** The start time gets the accent because it is the detail people act on. The callout closes the page because it is advice, not event data.

**Test:** I loaded the built page at phone and desktop widths and confirmed the three cells and the table fit without sideways scrolling.
