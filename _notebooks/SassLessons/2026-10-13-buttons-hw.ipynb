## Design Thinking

**Empathize:** A student opening this page may be stressed, so finding the right action has to take no effort.

**Define:** Every action needs a visual priority that matches its urgency, without raw CSS.

**Ideate:** Color could carry meaning instead of decoration, so I reserved red for anything that leads to urgent help.

**Prototype:** `<button>` runs an action on the page, and `<a>` goes somewhere else. `fill` marks the top priority and `outline` marks support options.

**Test:** In every bar I checked that one action stands out first and that each link opens the My Good Brain resources page.
