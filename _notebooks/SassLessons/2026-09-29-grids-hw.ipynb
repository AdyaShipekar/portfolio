## Reading a Lab Tray

A lab tray is already a grid, so beaker data is a natural fit for OCS grid classes. Each hack below replaces hand-built layout with the grid grammar.

### Popcorn Hack 1: Beaker Row

```python
%%html
<!-- UI_RUNNER: Grids Popcorn 1 | show_code: true -->
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell">Beaker A: Water</div>
  <div class="ocs__grid-cell">Beaker B: Salt Water</div>
  <div class="ocs__grid-cell">Beaker C: Vinegar</div>
</div>
```

### Popcorn Hack 2: Sizing with fr

Four `fr` values make four boxes, and `2fr` makes the Control Beaker twice as wide as each of the others.

```python
%%html
<!-- UI_RUNNER: Grids Popcorn 2 | show_code: true -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">
  <div class="ocs__grid-cell">Beaker 1</div>
  <div class="ocs__grid-cell">Beaker 2</div>
  <div class="ocs__grid-cell">Beaker 3</div>
  <div class="ocs__grid-cell">Control Beaker</div>
</div>
```

### Homework Hack: Readings and Keypad

The checklist runs in the browser and reads the grids on this page.

```python
%%html
<!-- UI_RUNNER: Grids Homework | show_code: true -->
<div id="lab-checks"></div>
<div id="lab-grids">
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>
  <div class="ocs__grid-cell">Beaker 1: 22.4</div>
  <div class="ocs__grid-cell">Beaker 2: 22.9</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>
</div>
<div class="ocs__grid ocs__grid--calculator">
  <div class="ocs__grid-cell">7</div>
  <div class="ocs__grid-cell">8</div>
  <div class="ocs__grid-cell">9</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>
  <div class="ocs__grid-cell">4</div>
  <div class="ocs__grid-cell">5</div>
  <div class="ocs__grid-cell">6</div>
  <div class="ocs__grid-cell ocs__grid-cell--muted">Min</div>
  <div class="ocs__grid-cell">1</div>
  <div class="ocs__grid-cell">2</div>
  <div class="ocs__grid-cell">3</div>
  <div class="ocs__grid-cell ocs__grid-cell--muted">Max</div>
</div>
</div>
<script>
let myHTML = document.getElementById('lab-grids').innerHTML;
let classList = [];
for (let found of (myHTML.match(/class="[^"]*"/g) || [])) {
  classList = classList.concat(found.slice(7, -1).split(' ').filter(c => c !== ''));
}
let madeUp = classList.filter(c => !c.startsWith('ocs__') && !c.startsWith('cols-'));

function line(label, ok) { return '<li>' + (ok ? 'PASS  ' : 'FIX  ') + label + '</li>'; }
document.getElementById('lab-checks').innerHTML = '<ul>' +
  line('No inline style attributes', !myHTML.includes('style=')) +
  line('No made up classes' + (madeUp.length > 0 ? ' (found: ' + madeUp.join(', ') + ')' : ''), madeUp.length === 0) +
  line('Has an ocs__grid root', classList.includes('ocs__grid')) +
  line('Uses the standard variant for Grid 1', classList.includes('ocs__grid--standard')) +
  line('Uses the calculator variant for Grid 2', classList.includes('ocs__grid--calculator')) +
  line('Every box uses ocs__grid-cell', classList.filter(c => c === 'ocs__grid-cell').length >= 8) +
  line('Title is a header cell', classList.includes('ocs__grid-cell--header')) +
  line('Has accent cells (outlier beaker and Avg key)', classList.filter(c => c === 'ocs__grid-cell--accent').length >= 2) +
  '</ul>';
</script>
```

## Design Thinking

**Empathize:** Someone reading lab results wants to spot the odd value first, not scan every number.

**Define:** The layout has to show structure and importance using only OCS classes.

**Ideate:** Cell types can act like labels. The header names the trial, and accent points at the reading to question.

**Prototype:** Beaker 3 gets the accent because 31.6 is far from the other two readings. I filled the keypad out to a full number pad, with Min and Max muted so Avg stays the main key.

**Test:** The homework checklist runs live on the page, and every line must show PASS.
