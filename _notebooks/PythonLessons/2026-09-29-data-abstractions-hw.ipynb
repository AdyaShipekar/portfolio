## Homework

```python
# CODE_RUNNER: Homework 
# REFACTOR: one list replaces separate star variables.
# Each element is one level's collected stars, in level order:
#   index 0 = Level 1, index 1 = Level 2, index 2 = Level 3
level_stars = [12, 18, 10]

# Add 5 stars to the second level, but only if a second level exists
# (skips the update instead of crashing with IndexError on short lists)
if len(level_stars) >= 2:
    level_stars[1] = level_stars[1] + 5

# Append 7 stars for a brand-new level at the end
level_stars.append(7)

# TRAVERSAL (Example C): visit every element and add it to the total (no sum())
total = 0
for stars in level_stars:
    total = total + stars

# Show the collection, the level count, and the total
# Expected: [12, 23, 10, 7] / 4 / 52
print(level_stars)
print(len(level_stars))
print(total)

# Show clear level labels so players never need to understand indices
for i in range(len(level_stars)):
    print(f"Level {i + 1}: {level_stars[i]} stars")

# TEST: total-only code with an empty list. Expected: 0 / 0
test_stars = []
# Guard skips the second-level update because there is no second element
if len(test_stars) >= 2:
    test_stars[1] = test_stars[1] + 5
test_total = 0
for stars in test_stars:
    test_total = test_total + stars
print(len(test_stars))
print(test_total)

# WHY A LIST: with 100 separate variables (level1_stars ... level100_stars),
# adding a level means creating a new variable name, and the total has to
# change by hand to include it. A count has to be updated by hand, too, and
# every print needs its own line. With a list, append() adds a level, and
# the same loop, len(), and label code work for any number of levels with
# zero edits.
```

## Design Thinking Process

**Need (player/maker):** Players want to see how many stars they earned on
each level and overall. Makers (developers) want to add levels without
rewriting the code every time.

**Goal:** Store every level's stars in one place, update or add levels
easily, and report the collection, level count, and total correctly for
any number of levels, including zero.

**Representations considered:**
- Separate variables (`level1_stars`, `level2_stars`, ...): easy to read for
  3 levels, but every new level needs a new name and edits to the total and
  count. This doesn't scale to 100 levels.
- One list (`level_stars`): elements are in level order, `append()` adds a
  level, and one loop handles the total. **Chosen.**

**Prototype:** Refactored the 3.1 prototype into `level_stars = [12, 18, 10]`,
added 5 to the second level, appended 7, and used a traversal loop for the
total.

**Actual test and revision:** Testing with an empty list showed that
`level_stars[1]` would raise an `IndexError` when there is no second level.
**Revision:** I wrapped the update in `if len(level_stars) >= 2:` so it is
skipped safely. After that, the empty-list test printed `0` and `0` as
expected.

**Player-facing improvement:** Players see "Level 1, Level 2, ..." instead of
indices, using `i + 1` as the label. Players don't need to know that lists
start at index 0.
