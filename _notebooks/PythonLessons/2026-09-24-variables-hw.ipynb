## Design Thinking

**Player's need:** Players want to know how they are doing in the game. They need to see how many stars they earned on each level, and a total that is always correct, even after bonuses change their score. They also need the game to feel comfortable to play, which means having some control over how it looks and moves on screen.

**Setting / representation I considered:** For motion, I used a Boolean (`True`/`False`) because the choice is simply on or off. I also considered a 0–100 motion slider, which is more flexible but harder to explain and test.

**Prototype I built:** This script. It stores the name, setting, and level stars, prints the total, adds a bonus to level two, and prints both the stored total and the recalculated total so any difference is visible.

**Why reduced motion should be a player choice:** Reduced motion should be a player choice because it allows the game experience to be personalized. Although some players enjoy the fast stars, others get dizzy, have headaches, or lose focus when there is a lot of screen motion, such as spinning stars, screen shake, or fast transitions. They still want to see their progress clearly. Turning the motion setting down doesn't change any of the game information or star totals.
