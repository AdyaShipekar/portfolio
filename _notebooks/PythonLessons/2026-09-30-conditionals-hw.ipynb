## Homework

```python
# CODE_RUNNER: Homework

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

# TODO 1: score each signal separately
# No else needed: if the message has no gift cards, there's nothing to add
if "gift card" in message.lower():
    risk_score += 4

# No else needed: a small amount leaves the score unchanged
if amount_requested > 100:
    risk_score += 3

# No else needed: a known contact adds no risk, so the score stays the same
if not known_contact:
    risk_score += 2

# TODO 2: report on the final score
# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print(f"Score {risk_score}: Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print(f"Score {risk_score}: This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print(f"Score {risk_score}: Be a little careful, but this is probably fine.")
else:
    print(f"Score {risk_score}: Nothing suspicious in this message.")
```

### Every branch is reachable

| Branch | message | amount_requested | known_contact | risk_score |
|---|---|---|---|---|
| >= 9 (scam) | "send gift cards now" | 500 | False | 4 + 3 + 2 = 9 |
| >= 6 (risky) | "send gift cards now" | 50 | False | 4 + 2 = 6 |
| >= 3 (careful) | "can you pay me back?" | 200 | True | 3 |
| else (clean) | "see you at lunch" | 0 | True | 0 |

The chain checks the highest threshold first. If it went in the order >= 3, >= 6, >= 9,
every score of 3 or more would stop at the first branch, and the other two would be dead code.

### Stretch: why `if` with no `else` is fine here

The scoring `if`s are independent signals that should each add points, and doing nothing when a signal is absent is the correct behavior. In Example A, the choices were mutually exclusive answers where exactly one had to run, so leaving out the `else` could leave a case with no answer at all.
