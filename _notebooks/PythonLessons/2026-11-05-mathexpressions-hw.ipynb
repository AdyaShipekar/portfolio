## Popcorn Hack 1

```python
# CODE_RUNNER: Popcorn Hack - Basic Arithmetic Operators
value_a = 100
value_b = 10
value_c = 25

# total of A, B, and C
addition = value_a + value_b + value_c

# difference between A and B
subtraction = value_a - value_b

# product of B and C
multiplication = value_b * value_c

# A divided by B ('/' always gives a float, so this prints 10.0)
division = value_a / value_b

print("Total:", addition)              # 135
print("Difference:", subtraction)      # 90
print("Product:", multiplication)      # 250
print("Division:", division)           # 10.0
```

## Popcorn Hack 2

```python
# CODE_RUNNER: Python Version of Popcorn Hack 2

# Grouping operations for the Poway Recovery Center booking system
total_monthly_capacity = 100
spots_requested_per_group = 4
sessions_booked = 3

# 1. Multiplication: total spots taken by the booking group
total_spots_taken = spots_requested_per_group * sessions_booked

# 2. Subtraction: remaining capacity for the center
remaining_capacity = total_monthly_capacity - total_spots_taken

# 3. Division: distribute remaining spots evenly across 2 active rooms
users_per_room = remaining_capacity / 2

print("Total Spots Taken:", total_spots_taken)     # 12
print("Remaining Capacity:", remaining_capacity)   # 88
print("Users per Room:", users_per_room)           # 44.0
```

## Homework

```python
# CODE_RUNNER: HOMEWORK

# Recovery Center Cost Report

cost_per_session = 35
sessions_per_week = 3
weeks_per_month = 4
enrollment_fee = 75   # one-time cost

# 1. Weekly cost: cost per session x sessions per week
weekly_cost = cost_per_session * sessions_per_week

# 2. Monthly cost: weekly cost x weeks per month
monthly_cost = weekly_cost * weeks_per_month

# 3. Total cost with fee: the fee is added once, not every week
total_cost_with_fee = monthly_cost + enrollment_fee

# 4. Average daily cost: monthly cost spread over 30 days ('/' gives a decimal)
average_daily_cost = monthly_cost / 30

# Print all results with labels
print("Weekly Cost:", weekly_cost)                          # 105
print("Monthly Cost:", monthly_cost)                        # 420
print("Total Cost with Fee:", total_cost_with_fee)          # 495
print("Average Daily Cost:", round(average_daily_cost, 2))  # 14.0
```
