## Popcorn Hack 2

```python
# CODE_RUNNER: Popcorn Hack 2
# Helper Procedure
def is_priority(incident_type):
    # Returns True only if the incident is a medical emergency
    return incident_type == "Medical Emergency"


# Outer Procedure
def process_report(location, incident_type):
    # Calls the helper and stores the True/False result
    priority = is_priority(incident_type)

    # Prints an urgent message if the incident is a priority
    if priority:
        print(f"PRIORITY: {incident_type} at {location}. Send help immediately!")
    # Prints a standard message for everything else
    else:
        print(f"Standard report: {incident_type} at {location}.")


# Tests the priority branch
process_report("Main Street", "Medical Emergency")
# Tests the standard branch
process_report("City Park", "Noise Complaint")
```

## Homework

```python
# CODE_RUNNER: Homework

# Helper Procedure
def check_priority(incident_type):
    # Returns True if the incident is one of the urgent types
    return incident_type in ["Medical Emergency", "Fire", "Car Accident"]


# Outer Procedure
def dispatch_response(location, incident_type):
    # Calls the helper and stores the True/False result
    priority = check_priority(incident_type)

    # Sends emergency responders if the incident is a priority
    if priority:
        print(f"URGENT: Dispatching emergency responders to {location} for a {incident_type}!")
    # Schedules a normal response for everything else
    else:
        print(f"Routine: An officer will follow up at {location} about a {incident_type}.")


# Tests a priority incident
dispatch_response("Main Street", "Fire")
# Tests another priority incident at a different location
dispatch_response("Highway 5", "Car Accident")
# Tests a non-priority incident
dispatch_response("City Park", "Noise Complaint")
```
