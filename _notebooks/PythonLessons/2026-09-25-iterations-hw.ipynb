## Popcorn Hack

```python
# CODE_RUNNER: Popcorn Hack
# Theme: The SFSRC team scans wildfire alert levels from different zones.
# Goal: skip safe zones, count response zones, total their alert scores, and stop at the first critical zone.

alerts = [1, 4, 2, 6, 3, 9, 5]

# Pick the cutoff for safe alerts. Alerts below this number should be skipped.
safe_cutoff = 4

# Pick the cutoff for critical alerts. The loop should stop at this number or higher.
critical_cutoff = 8

# Counters and totals start at 0 because no alerts have been processed yet.
action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

# Loop through the alert list.
for alert in alerts:
    # If the alert is safe, count it, print a message, and skip to the next alert.
    if alert < safe_cutoff:
        safe_count += 1
        print("Alert " + str(alert) + ": safe, skipping")
        continue

    # If code reaches here, the alert needs action.
    # Count this action alert and add its score to the running total.
    action_count += 1
    total_action_score += alert
    print("Alert " + str(alert) + ": action needed")

    # If this alert is critical, mark it and stop the scan.
    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found - stopping scan")
        break

# Final report: these lines should explain what happened during the scan.
print("Safe alerts skipped: " + str(safe_count))
print("Action alerts: " + str(action_count))
print("Total action score: " + str(total_action_score))
print("Critical found: " + str(critical_found))
```

## Homework

```python
# CODE_RUNNER: Homework Hack - themed loop-based checker
# Theme: Server load monitor for a school web server.
# Each number is the server's load as a percent of its capacity.
# Goal: skip safe loads, count loads that need attention, total their scores,
# and stop at the first critical overload.

server_loads = [35, 62, 48, 71, 20, 55, 93, 40, 80]

# Loads below safe_cutoff are normal and should be skipped with continue.
safe_cutoff = 50

# Loads at or above critical_cutoff mean the server is overloaded, so the scan stops with break.
critical_cutoff = 90

# Summary variables
checked_count = 0
action_count = 0
skipped_count = 0
total_score = 0
critical_found = False

# Traverse every load in the list until a critical load stops the scan.
for load in server_loads:
    # Count every load the loop looks at.
    checked_count += 1

    # Skip safe loads. Count them before continuing.
    if load < safe_cutoff:
        skipped_count += 1
        print(str(load) + "%: safe, no action")
        continue

    # If code reaches here, this load needs action.
    # Count it and add it to the running total.
    action_count += 1
    total_score += load
    print(str(load) + "%: action needed")

    # Stop early if this load is critical.
    if load >= critical_cutoff:
        critical_found = True
        print("Critical server overload found. Stop checking.")
        break

# Final report
print("Final Server Report")
print("Loads checked: " + str(checked_count))
print("Safe loads skipped: " + str(skipped_count))
print("Loads needing action: " + str(action_count))
print("Total action load: " + str(total_score))
print("Critical overload found: " + str(critical_found))
```
