## Homework

```python
# CODE_RUNNER: Homework
# PREDICTIONS (before running):
#   n = 12: books 12, visits 12, ordered pairs 12 x 12 = 144
#   n = 24: books 24, visits 24, ordered pairs 24 x 24 = 576
#   n = 0:  books 0,  visits 0,  ordered pairs 0
#   1,000 books: linear worst case 1000, binary worst case floor(log2(1000)) + 1 = 10

import math

# Report the work needed for a catalog of n books (n is a nonnegative whole number)
def catalog_report(n):
    # Visits: look at each book once, so the count grows at the same rate as n
    visits = 0
    for book in range(n):
        visits = visits + 1

    # Ordered pairs: compare every book with every book (nested loops), so n x n
    pairs = 0
    for first in range(n):
        for second in range(n):
            pairs = pairs + 1

    # Clear labels so the librarian doesn't need to know the formulas
    print(f"Books in catalog: {n}")
    print(f"Visits (look at each book once): {visits}")
    print(f"Ordered pairs (compare every book with every book): {pairs}")
    print()

# Main run: 12 books
catalog_report(12)

# Doubled catalog: 24 books
catalog_report(24)

# Empty catalog test: 0 books (the loops run zero times, so nothing crashes)
catalog_report(0)

# Checkpoint B: worst-case checks for searching 1,000 books
big_n = 1000
linear_checks = big_n
# Guard against n = 0, because log2(0) is undefined
binary_checks = 0 if big_n == 0 else math.floor(math.log2(big_n)) + 1
print(f"Linear search worst case for {big_n} books: {linear_checks} checks")
print(f"Binary search worst case for {big_n} books: {binary_checks} checks")
```

# Growth Rates (12 → 24 books)

Visits doubled (12 → 24) because visiting each book once is linear growth:
twice the books means twice the work. Ordered pairs quadrupled (144 → 576)
because every book is paired with every book (n × n). Doubling n gives
2 × 2 = 4 times the work. This quadratic growth gets slow very quickly as
the catalog grows.

# Recommendation (1,000 books)

For repeated searches of an already sorted catalog, use **binary search**:
at most 10 checks, compared with 1,000 for linear search. That's about
100 times fewer checks per search, and the savings add up with every
repeated search.

**If the catalog is unsorted**, binary search doesn't work, because it relies
on order to throw away half the list. You would need to sort first, which
costs extra work. For a single search, plain linear search (1,000 checks)
is simpler. Sorting pays off only if you will search many times.

**If you must store an extra sorted copy**, you use twice the memory, and
both copies must stay in sync whenever books are added or removed. This
trade is worth it for a catalog searched often and changed rarely. It is
worth less for one that changes constantly.

# Design Thinking Process

**Librarian need:** The librarian needs to know how much work different
catalog tasks take as the collection grows. The librarian shouldn't have to
read formulas or code to get that answer.

**Goal:** Report the book count, the visits, and the ordered pairs for
any catalog size, including an empty one. Then recommend a search method
backed by real counts.

**Methods considered:**
- Linear search: works on any catalog, sorted or not, but the worst case
  checks every book.
- Binary search: very few checks (10 for 1,000 books), but only works on a
  sorted catalog.
- Hard-coding the answers (`print(144)`): rejected, because it breaks as soon
  as the catalog size changes.

**Prototype:** A `catalog_report(n)` function counts visits with one loop and
ordered pairs with nested loops. It is called for 12, 24, and 0 books,
followed by a checkpoint B comparison for 1,000 books.

**Actual test and revision:** I tested the binary-search formula without the

**Revision:** I added `0 if n == 0 else ...` so an empty catalog returns
0 checks instead of crashing. The first draft also printed bare numbers,
like `144`, which a librarian couldn't interpret. **Revision:** I added
plain-language labels, such as "Ordered pairs (compare every book with
every book)", so the output explains itself.
