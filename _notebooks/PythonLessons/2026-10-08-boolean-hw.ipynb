## Popcorn Hack 3

```python
# CODE_RUNNER: Popcorn Hack 3 - SFI Search Filter

# Prediction (match):    query "1.2" matches "Multiple Disc Clutch Assemblies" by exact spec number
# Prediction (nonmatch): "Replacement Flywheels and Clutch Assemblies" won't match, since its spec is "1.1"

query = "1.2"                  # the search text the user typed

records = [
    {
        "product_name": "Replacement Flywheels and Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

valid_categories = ["Auto Racing", "Drag Racing"]   # only these categories can appear in results

results = []                                         # empty list to collect matching records

for record in records:                               # check each record one at a time
    name_match = query.lower() in record["product_name"].lower()   # True if query is inside the name (ignores case)
    spec_match = query == record["spec_number"]                     # True only if query is the exact spec number
    category_ok = record["category"] in valid_categories            # True if the category is accepted

    if (name_match or spec_match) and category_ok:   # 'or' allows either match, 'and' requires a valid category
        results.append(record)                       # keep this record

for record in results:                               # print each valid match
    print(record["spec_number"], "-", record["product_name"], "-", record["category"])

if not results:                                      # 'not' on an empty list is True
    print("No matching records")

# Output:
# 1.2 - Multiple Disc Clutch Assemblies - Drag Racing

# Change: query = "flywheel"
# New output:
# 1.1 - Replacement Flywheels and Clutch Assemblies - Auto Racing
# 2.1 - Racing Flywheel Record - Auto Racing
# Why it changed: "flywheel" isn't a spec number, but it appears in two product
# names, so name_match becomes True for those records and both are in valid categories

# Change: set the second record's category to "Karting" and query = "1.2"
# New output:
# No matching records
# Why it changed: the spec still matches, but category_ok is False,
# so the 'and' blocks the record from the results
```

```python
# CODE_RUNNER: Homework Hack - SFI Backend Validator

existing_spec_numbers = ["1.1", "2.1"]               # spec numbers already stored in the backend
valid_categories = ["Auto Racing", "Drag Racing"]    # the only accepted racing categories

test_records = [
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.2",
        "effective_date": "Feb. 9, 2006"
    },
    {
        "product_name": "Replacement Flywheels",
        "category": "Street Car",
        "spec_number": "3.1",
        "effective_date": "Jan. 1, 2026"
    },
    {
        "product_name": "Existing Flywheel Record",
        "category": "Drag Racing",
        "spec_number": "1.1",
        "effective_date": "Jan. 1, 2026"
    },
    {
        "product_name": "",                          # extra test: blank product name
        "category": "Drag Racing",
        "spec_number": "4.1",
        "effective_date": ""                         # extra test: missing effective date
    }
]

# Reusable validator: works for any record, category list, or stored spec list
def validate_record(record, valid_categories, existing_spec_numbers):
    has_product_name = record.get("product_name", "").strip() != ""      # name is not blank
    valid_category = record.get("category", "") in valid_categories      # category is accepted
    has_spec_number = record.get("spec_number", "").strip() != ""        # spec number is not blank
    has_effective_date = record.get("effective_date", "").strip() != ""  # date is not blank
    duplicate_spec = record.get("spec_number", "") in existing_spec_numbers  # spec already stored?

    # 'and' requires every check to pass, 'not' requires the spec to be new
    return (
        has_product_name
        and valid_category
        and has_spec_number
        and has_effective_date
        and not duplicate_spec
    )

# Test every record and print the decision
for record in test_records:                                              # loop through each candidate
    accepted = validate_record(record, valid_categories, existing_spec_numbers)  # run the validator
    status = "ACCEPTED" if accepted else "REJECTED"                      # turn the Boolean into a label
    name = record["product_name"] or "(no name)"                         # show a placeholder if blank
    print(status, "-", record["spec_number"], "-", name)                 # print the result

# Output:
# ACCEPTED - 1.2 - Multiple Disc Clutch Assemblies
# REJECTED - 3.1 - Replacement Flywheels        (Street Car is not an accepted category)
# REJECTED - 1.1 - Existing Flywheel Record     (spec 1.1 already exists in the backend)
# REJECTED - 4.1 - (no name)                    (blank product name and missing effective date)

# Change: add "Street Car" to valid_categories
# New output for record 2:
# ACCEPTED - 3.1 - Replacement Flywheels
# Why it changed: the category check now passes and every other field was already valid
```
